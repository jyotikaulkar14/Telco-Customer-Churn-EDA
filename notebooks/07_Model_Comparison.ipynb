```python
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
```

```python
import joblib

X_test = pd.read_csv("../data/X_test_scaled.csv")
y_test = pd.read_csv("../data/y_test.csv").squeeze()

logistic_model = joblib.load("../data/logistic_regression_model.pkl")
decision_tree_model = joblib.load("../data/decision_tree_model.pkl")
random_forest_model = joblib.load("../data/random_forest_model.pkl")

print("Test data and models loaded successfully!")
```

```python
logistic_pred = logistic_model.predict(X_test)
decision_tree_pred = decision_tree_model.predict(X_test)
random_forest_pred = random_forest_model.predict(X_test)

logistic_prob = logistic_model.predict_proba(X_test)[:, 1]
decision_tree_prob = decision_tree_model.predict_proba(X_test)[:, 1]
random_forest_prob = random_forest_model.predict_proba(X_test)[:, 1]

print("Predictions generated successfully!")
```

```python
models = {
    "Logistic Regression": (logistic_pred, logistic_prob),
    "Decision Tree": (decision_tree_pred, decision_tree_prob),
    "Random Forest": (random_forest_pred, random_forest_prob)
}

results = []

for model_name, (predictions, probabilities) in models.items():
    
    results.append({
        "Model": model_name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(y_test, predictions),
        "Recall": recall_score(y_test, predictions),
        "F1-Score": f1_score(y_test, predictions),
        "ROC-AUC": roc_auc_score(y_test, probabilities)
    })

results_df = pd.DataFrame(results)

results_df = results_df.round(4)

display(results_df)
```

```python
best_model = results_df.loc[
    results_df["F1-Score"].idxmax()
]

print("🏆 BEST MODEL")
print("------------------------")
print("Model:", best_model["Model"])
print("F1-Score:", best_model["F1-Score"])
print("Accuracy:", best_model["Accuracy"])
print("Precision:", best_model["Precision"])
print("Recall:", best_model["Recall"])
print("ROC-AUC:", best_model["ROC-AUC"])
```

```python
results_df.to_csv(
    "../data/model_comparison_results.csv",
    index=False
)

print("Model comparison results saved successfully!")
```
