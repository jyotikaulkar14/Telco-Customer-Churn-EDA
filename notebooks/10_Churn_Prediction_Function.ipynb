```python
import pandas as pd
import joblib

# Load trained model
model = joblib.load("../data/best_churn_model.pkl")

# Load scaler
scaler = joblib.load("../data/standard_scaler.pkl")

# Load training data to get exact feature structure
X_train = pd.read_csv("../data/X_train.csv")

feature_names = X_train.columns.tolist()

print("Model loaded successfully!")
print("Scaler loaded successfully!")
print("Number of features:", len(feature_names))
```

```python
def predict_churn(customer_data):
    
    # Convert customer information into DataFrame
    customer_df = pd.DataFrame([customer_data])
    
    # Make sure columns match training data
    customer_df = customer_df.reindex(
        columns=feature_names,
        fill_value=0
    )
    
    # Scale the customer data
    customer_scaled = scaler.transform(customer_df)
    
    # Make prediction
    prediction = model.predict(customer_scaled)[0]
    
    # Get churn probability
    probability = model.predict_proba(customer_scaled)[0][1]
    
    # Display result
    print("======================================")
    print("       CUSTOMER CHURN PREDICTION")
    print("======================================")
    
    if prediction == 1:
        print("Prediction: YES - Customer likely to churn")
    else:
        print("Prediction: NO - Customer likely to stay")
    
    print(f"Churn Probability: {probability * 100:.2f}%")
    print("======================================")
    
    return prediction, probability
```
