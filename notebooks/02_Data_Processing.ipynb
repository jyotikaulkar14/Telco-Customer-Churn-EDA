```python
import pandas as pd

df = pd.read_excel("../data/Customer_Churn_Dataset.xlsx")

# Convert TotalCharges from object to numeric
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

# Check missing values created during conversion
print("Missing TotalCharges:", df["TotalCharges"].isna().sum())

# Replace missing TotalCharges with 0
df["TotalCharges"] = df["TotalCharges"].fillna(0)

print("TotalCharges datatype:", df["TotalCharges"].dtype)
```

```python
print(df["TotalCharges"].dtype)
print(df["TotalCharges"].head())
```

```python
# Separate features and target
# Separate features and target

X = df.drop(columns=["customerID", "Churn"])
y = df["Churn"].map({"No": 0, "Yes": 1})

print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nCategorical columns:")
print(X.select_dtypes(include=["object", "string", "category"]).columns.tolist())

print("\nNumerical columns:")
print(X.select_dtypes(include=["number"]).columns.tolist())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Identify categorical and numerical columns
categorical_cols = X_train.select_dtypes(
    include=["object", "string", "category"]
).columns.tolist()

numerical_cols = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

print("Categorical columns:", categorical_cols)
print("\nNumerical columns:", numerical_cols)

# Create preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_cols),
        ("cat", OneHotEncoder(
            handle_unknown="ignore",
            drop="first",
            sparse_output=False
        ), categorical_cols)
    ],
    verbose_feature_names_out=False
)

# Make the output a pandas DataFrame
preprocessor.set_output(transform="pandas")

print("\nPreprocessor created successfully!")
```

```python
# Fit preprocessing only on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Apply the same preprocessing to testing data
X_test_processed = preprocessor.transform(X_test)

print("Processed training data shape:", X_train_processed.shape)
print("Processed testing data shape:", X_test_processed.shape)
```

```python
import joblib

# Save processed datasets
X_train_processed.to_csv("../data/X_train_processed.csv", index=False)
X_test_processed.to_csv("../data/X_test_processed.csv", index=False)

# Save target datasets
y_train.to_csv("../data/y_train_processed.csv", index=False)
y_test.to_csv("../data/y_test_processed.csv", index=False)

# Save the fitted preprocessing pipeline
joblib.dump(preprocessor, "../data/preprocessor.pkl")

print("Processed datasets saved successfully!")
print("Preprocessor saved successfully!")
```
