```python
import pandas as pd
import joblib

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
X_train = pd.read_csv("../data/X_train_scaled.csv")
X_test = pd.read_csv("../data/X_test_scaled.csv")

y_train = pd.read_csv("../data/y_train.csv").squeeze()
y_test = pd.read_csv("../data/y_test.csv").squeeze()

print("Data loaded successfully!")

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

logistic_model.fit(X_train, y_train)

print("Logistic Regression trained successfully!")
```

```python
decision_tree_model = DecisionTreeClassifier(
    random_state=42,
    max_depth=5
)

decision_tree_model.fit(X_train, y_train)

print("Decision Tree trained successfully!")
```

```python
random_forest_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    max_depth=10,
    n_jobs=-1
)

random_forest_model.fit(X_train, y_train)

print("Random Forest trained successfully!")
```

```python
logistic_pred = logistic_model.predict(X_test)
decision_tree_pred = decision_tree_model.predict(X_test)
random_forest_pred = random_forest_model.predict(X_test)

print("Predictions generated successfully!")
```

```python
logistic_prob = logistic_model.predict_proba(X_test)[:, 1]

decision_tree_prob = decision_tree_model.predict_proba(X_test)[:, 1]

random_forest_prob = random_forest_model.predict_proba(X_test)[:, 1]

print("Prediction probabilities generated successfully!")
```

```python
joblib.dump(
    logistic_model,
    "../data/logistic_regression_model.pkl"
)

joblib.dump(
    decision_tree_model,
    "../data/decision_tree_model.pkl"
)

joblib.dump(
    random_forest_model,
    "../data/random_forest_model.pkl"
)

print("All models saved successfully!")
```
